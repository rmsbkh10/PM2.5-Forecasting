```python

# ============================================================
# PM2.5 FORECASTING - PART A: DATA EXPLORATION
# Adapted for VS Code
# ============================================================

# ============================================================
# IMPORTS
# ============================================================

import pandas as pd
import numpy as np
import os
import zipfile
import json
import warnings

from pathlib import Path
from IPython.display import display

import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats

from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.model_selection import TimeSeriesSplit, GroupKFold, KFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tsa.stattools import adfuller, kpss, acf, pacf
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.stats.diagnostic import acorr_ljungbox
from statsmodels.tools.sm_exceptions import InterpolationWarning

warnings.filterwarnings("ignore")
warnings.simplefilter("ignore", InterpolationWarning)

np.random.seed(42)

# ============================================================
# PROJECT CONFIGURATION
# ============================================================

CURRENT = Path.cwd()

ROOT = next(
    (p for p in [CURRENT, *CURRENT.parents] if (p / "data").is_dir()),
    CURRENT.parent if CURRENT.name == "notebooks" else CURRENT
)

DATA_PATH = ROOT / "data" / "raw" / "combined_gee_pm25.csv"

GROUPE = "G22"

OUT = ROOT / "results" / GROUPE
FIGURES = ROOT / "figures"

OUT.mkdir(parents=True, exist_ok=True)
FIGURES.mkdir(parents=True, exist_ok=True)

print("PROJECT ROOT:", ROOT)
print("DATASET PATH:", DATA_PATH)

# ============================================================
# ETAPE 1 : IMPORTER LE FICHIER
# ============================================================

print("\nETAPE 1 : Chargement du dataset")

# Si le CSV n'existe pas, rechercher une archive ZIP
if not DATA_PATH.exists():

    zip_files = list(DATA_PATH.parent.glob("*.zip"))

    if zip_files:

        extract_folder = DATA_PATH.parent / "extracted"
        extract_folder.mkdir(parents=True, exist_ok=True)

        for zip_file in zip_files:

            with zipfile.ZipFile(zip_file, "r") as archive:
                archive.extractall(extract_folder)

        matches = list(
            extract_folder.rglob("combined_gee_pm25.csv")
        )

        if matches:
            DATA_PATH = matches[0]

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset introuvable : {DATA_PATH}\n"
        "Place combined_gee_pm25.csv dans data/raw/"
    )

print("Fichier trouvé :", DATA_PATH)

# ============================================================
# ETAPE 2 : CHARGER LES DONNEES
# ============================================================

df = pd.read_csv(DATA_PATH)

print("\n========== DATASET OVERVIEW ==========")
print("Shape :", df.shape)

print("\nColonnes :")
print(df.columns.tolist())

# ============================================================
# ETAPE 3 : PREMIERES LIGNES
# ============================================================

print("\n========== FIRST 5 ROWS ==========")
display(df.head())

# ============================================================
# ETAPE 4 : INFORMATIONS GENERALES
# ============================================================

print("\n========== DATA INFORMATION ==========")
df.info()

# ============================================================
# ETAPE 5 : VALEURS MANQUANTES
# ============================================================

print("\n========== MISSING VALUES ==========")
display(df.isnull().sum())

# ============================================================
# ETAPE 6 : STATISTIQUES DESCRIPTIVES
# ============================================================

print("\n========== DESCRIPTIVE STATISTICS ==========")
display(df.describe(include="all"))

# ============================================================
# ETAPE 7 : DOUBLONS
# ============================================================

print("\n========== DUPLICATES ==========")
print("Nombre de doublons :", df.duplicated().sum())

# ============================================================
# ETAPE 8 : CONFIGURATION
# ============================================================

print("\nETAPE 8 : Configuration terminée")

# ============================================================
# ETAPE 9 : DETECTION DES COLONNES
# ============================================================

def trouver(mots):
    for c in df.columns:
        if any(m in c.lower() for m in mots):
            return c
    return None

DATE = trouver(["date", "time"])
STATION = "sensor_name"
LAT = trouver(["lat"])
LON = trouver(["lon", "lng"])
TARGET = trouver(["pm2.5", "pm25", "pm2_5"])

required = {
    "DATE": DATE,
    "STATION": STATION if STATION in df.columns else None,
    "LAT": LAT,
    "LON": LON,
    "TARGET": TARGET
}

missing_columns = [
    k for k, v in required.items() if v is None
]

if missing_columns:
    raise ValueError(
        f"Colonnes introuvables : {missing_columns}\n"
        f"Colonnes disponibles : {df.columns.tolist()}"
    )

print("Date:", DATE)
print("Station:", STATION)
print("Latitude:", LAT)
print("Longitude:", LON)
print("Cible:", TARGET)

df[DATE] = pd.to_datetime(
    df[DATE], errors="coerce"
).dt.normalize()

# ============================================================
# ETAPE 10 : AUDIT DU DATASET
# ============================================================

print("\n========== DATASET AUDIT ==========")

print("Période :", df[DATE].min(), "->", df[DATE].max())
print("Nombre de stations :", df[STATION].nunique())

print("Latitude :", df[LAT].min(), "->", df[LAT].max())
print("Longitude :", df[LON].min(), "->", df[LON].max())

print("Dates illisibles :", df[DATE].isna().sum())

print(
    "Stations avec plusieurs coordonnées :",
    (df.groupby(STATION)[[LAT, LON]].nunique() > 1)
    .any(axis=1).sum()
)

print(
    "Mesures par station :",
    df.groupby(STATION).size().agg(
        ["min", "median", "max"]
    ).tolist()
)

# Détection des variables de fuite
FUITE = [
    c for c in df.columns
    if c != TARGET and any(
        m in c.lower()
        for m in ["pm10", "aqi", "pm2", "pm25"]
    )
]

print("Variables exclues :", FUITE)

EXCLURE = [DATE, STATION, LAT, LON, TARGET] + FUITE

PRED = [
    c for c in df.select_dtypes("number").columns
    if c not in EXCLURE and "id" not in c.lower()
]

if not PRED:
    raise ValueError("Aucun prédicteur numérique disponible.")

print("Prédicteurs :", PRED)

# ============================================================
# ETAPE 11 : NETTOYAGE DE LA CIBLE ET DOUBLONS
# ============================================================

n0 = len(df)

n_manquant = df[TARGET].isna().sum()
n_bas = (df[TARGET] <= 0).sum()
n_haut = (df[TARGET] >= 500).sum()

df = df[
    df[TARGET].notna()
    & (df[TARGET] > 0)
    & (df[TARGET] < 500)
    & df[DATE].notna()
].copy()

print(
    f"Supprimées : {n_manquant} manquantes, "
    f"{n_bas} <= 0, {n_haut} >= 500"
)

print(f"Lignes : {n0} -> {len(df)}")

n_dup_total = df.duplicated().sum()

n_dup_cle = df.duplicated(
    [DATE, LAT, LON, STATION]
).sum()

df = df.drop_duplicates()

df = df.groupby(
    [STATION, DATE],
    as_index=False
).agg({
    c: "mean" if pd.api.types.is_numeric_dtype(df[c])
    else "first"
    for c in df.columns
    if c not in [STATION, DATE]
})

print("Doublons complets :", n_dup_total)
print("Doublons sur la clé :", n_dup_cle)

# ============================================================
# ETAPE 12 : STATISTIQUES DESCRIPTIVES
# ============================================================

VARS = [TARGET] + PRED

stats_df = df[VARS].describe().T

stats_df["skewness"] = df[VARS].skew()
stats_df["kurtosis"] = df[VARS].kurt()

display(stats_df.round(3))

# ============================================================
# ETAPE 13 : GRAPHIQUES DES VARIABLES CLES
# ============================================================

CLES = [TARGET] + [
    c for c in PRED
    if any(m in c.lower() for m in [
        "aod", "no2", "co", "o3", "temp",
        "press", "wind", "u_", "v_"
    ])
][:8]

fig, ax = plt.subplots(
    2, len(CLES),
    figsize=(3 * len(CLES), 6),
    squeeze=False
)

for i, c in enumerate(CLES):

    sns.histplot(
        df[c].dropna(),
        bins=50,
        ax=ax[0, i]
    )

    ax[0, i].set_title(c, fontsize=9)

    sns.boxplot(
        y=df[c].dropna(),
        ax=ax[1, i]
    )

plt.tight_layout()
plt.savefig(FIGURES / "key_variables.png", bbox_inches="tight")
plt.show()

df["mois"] = df[DATE].dt.month

df["saison"] = df["mois"].map({
    12: "Hiver", 1: "Hiver", 2: "Hiver",
    3: "Printemps", 4: "Printemps", 5: "Printemps",
    6: "Eté", 7: "Eté", 8: "Eté",
    9: "Automne", 10: "Automne", 11: "Automne"
})

plt.figure(figsize=(10, 5))

sns.boxplot(data=df, x="mois", y=TARGET)

plt.title("PM2.5 par mois")
plt.savefig(FIGURES / "pm25_by_month.png", bbox_inches="tight")
plt.show()

# ============================================================
# ETAPE 14 : VALEURS MANQUANTES ET OUTLIERS
# ============================================================

manq = (
    100 * df[VARS].isna().mean()
).sort_values(ascending=False)

print("% de valeurs manquantes :")
display(manq.round(2))

lignes = []

for c in VARS:

    x = df[c].dropna()

    if len(x) < 2:
        continue

    q1, q3 = x.quantile([.25, .75])
    iqr = q3 - q1

    ech = x.sample(
        min(len(x), 20000),
        random_state=42
    ).values.reshape(-1, 1)

    if len(ech) < 2:
        continue

    iso = IsolationForest(
        random_state=42
    ).fit(ech)

    lof_neighbors = min(20, len(ech) - 1)

    lof = LocalOutlierFactor(
        n_neighbors=lof_neighbors
    ).fit_predict(ech)

    z = np.abs(
        stats.zscore(x, nan_policy="omit")
    )

    lignes.append({
        "variable": c,

        "%_IQR": 100 * (
            (x < q1 - 1.5 * iqr)
            | (x > q3 + 1.5 * iqr)
        ).mean(),

        "%_zscore": 100 * (z > 3).mean(),

        "%_IsolationForest": 100 * (
            iso.predict(ech) == -1
        ).mean(),

        "%_LOF": 100 * (lof == -1).mean()
    })

outliers = pd.DataFrame(lignes).set_index("variable")

display(outliers.round(2))

# ============================================================
# ETAPE 15 : CORRELATIONS ET VIF
# ============================================================

fig, ax = plt.subplots(1, 2, figsize=(16, 6))

sns.heatmap(
    df[VARS].corr(method="pearson"),
    cmap="coolwarm",
    center=0,
    ax=ax[0]
)

ax[0].set_title("Pearson")

sns.heatmap(
    df[VARS].corr(method="spearman"),
    cmap="coolwarm",
    center=0,
    ax=ax[1]
)

ax[1].set_title("Spearman")

plt.tight_layout()
plt.show()

corr = df[VARS].corr().abs()

paires = [
    (a, b, round(corr.loc[a, b], 2))
    for i, a in enumerate(VARS)
    for b in VARS[i + 1:]
    if corr.loc[a, b] > 0.9
]

print("Paires très corrélées :", paires)

X = df[PRED].copy()

X = X.fillna(X.median())

X = X.loc[:, X.nunique() > 1]

X = (X - X.mean()) / X.std()

if X.shape[1] >= 2:

    vif = pd.Series(
        [
            variance_inflation_factor(X.values, i)
            for i in range(X.shape[1])
        ],
        index=X.columns
    )

else:
    vif = pd.Series(dtype=float)

print("VIF (> 10 = colinéarité forte) :")
display(vif.round(2))

# ============================================================
# ETAPE 16 : PM2.5 VS PREDICTEURS
# ============================================================

print("Corrélation de Spearman par saison :")

display(
    df.groupby("saison").apply(
        lambda g: g[PRED].corrwith(
            g[TARGET],
            method="spearman"
        )
    ).T.round(2)
)

y = df[TARGET]

print(
    f"Jours > 25 µg/m³ : "
    f"{100 * (y > 25).mean():.1f}%"
)

print(
    f"Jours > 50 µg/m³ : "
    f"{100 * (y > 50).mean():.1f}%"
)

print("Skewness PM2.5 :", round(y.skew(), 2))

print(
    "Skewness après log1p :",
    round(np.log1p(y).skew(), 2)
)

# ============================================================
# ETAPE 17 : EXPLORATORY_DATA.CSV
# ============================================================

rows = []

for c in VARS:

    x = df[c]
    partner_corr = corr[c].drop(c)

    r = {
        "model_id": "arima_woa_lstm_pm25",
        "model_group": GROUPE,
        "variable": c,
        "type": "numeric",
        "n_total": len(x),
        "n_missing": x.isna().sum(),
        "pct_missing": 100 * x.isna().mean(),
        "n_unique": x.nunique(),
        "n_duplicates_key": n_dup_cle,
        "min": x.min(),
        "q1": x.quantile(.25),
        "median": x.median(),
        "q3": x.quantile(.75),
        "max": x.max(),
        "mean": x.mean(),
        "std": x.std(),
        "skewness": x.skew(),
        "kurtosis": x.kurt(),
        "iqr": x.quantile(.75) - x.quantile(.25),
        "pct_out_iqr": outliers.loc[c, "%_IQR"],
        "pct_out_z": outliers.loc[c, "%_zscore"],
        "pct_out_if": outliers.loc[c, "%_IsolationForest"],
        "pct_out_lof": outliers.loc[c, "%_LOF"],
        "max_abs_corr": partner_corr.max() if len(partner_corr) else np.nan,
        "corr_partner": partner_corr.idxmax() if len(partner_corr) else None,
        "vif": vif.get(c, np.nan),
        "is_leakage_suspect": False
    }

    if c == TARGET:

        r.update({
            "family": "Target",
            "unit": "µg/m³",
            "n_sentinel_low": n_bas,
            "n_sentinel_high": n_haut,
            "n_removed_total": n0 - len(df) + n_dup_total
        })

    else:

        ok = df[[c, TARGET]].dropna()

        if len(ok) >= 2:
            rho, p = stats.spearmanr(ok[c], ok[TARGET])
        else:
            rho, p = np.nan, np.nan

        r.update({
            "target_assoc": f"spearman={rho:.3f}",
            "target_pval": p
        })

    rows.append(r)

exploratory_data = pd.DataFrame(rows)

exploratory_data.to_csv(
    OUT / "Exploratory_Data.csv",
    index=False
)

print("Sauvegardé :", OUT / "Exploratory_Data.csv")

# ============================================================
# ETAPE 18 : ANALYSE TEMPORELLE
# ============================================================

serie = (
    df.groupby(DATE)[TARGET]
    .median()
    .sort_index()
    .asfreq("D")
)

serie_clean = serie.dropna()

if len(serie_clean) < 5:
    raise ValueError("Pas assez de données temporelles.")

fig, ax = plt.subplots(1, 2, figsize=(14, 4))

plot_acf(
    serie_clean,
    lags=min(40, len(serie_clean) // 2 - 1),
    ax=ax[0]
)

plot_pacf(
    serie_clean,
    lags=min(20, len(serie_clean) // 2 - 1),
    ax=ax[1],
    method="ywm"
)

plt.tight_layout()
plt.show()

tempo = []

for st, g in df.groupby(STATION):

    s = g.set_index(DATE)[TARGET].sort_index()

    full_dates = pd.date_range(
        s.index.min(),
        s.index.max(),
        freq="D"
    )

    s = s.reindex(full_dates)

    x = s.dropna()

    if len(x) < 60:
        continue

    a = acf(
        s.values,
        nlags=10,
        missing="conservative"
    )

    pente = stats.linregress(
        (x.index - x.index[0]).days,
        x.values
    )

    try:
        pacf_lag1 = pacf(x, nlags=2)[1]
    except Exception:
        pacf_lag1 = np.nan

    try:
        ljung_p = acorr_ljungbox(
            x, lags=[7]
        )["lb_pvalue"].iloc[0]
    except Exception:
        ljung_p = np.nan

    try:
        adf_p = adfuller(x)[1]
    except Exception:
        adf_p = np.nan

    try:
        kpss_p = kpss(x, nlags="auto")[1]
    except Exception:
        kpss_p = np.nan

    seasonal = x.groupby(x.index.month).mean()

    tempo.append({
        "model_id": "arima_woa_lstm_pm25",
        "model_group": GROUPE,
        "variable": TARGET,
        "node_id": st,
        "n_records": len(x),
        "t_min": x.index.min().date(),
        "t_max": x.index.max().date(),

        "sampling_interval_median_sec":
            x.index.to_series().diff()
            .dt.total_seconds().median(),

        "pct_missing_intervals": 100 * s.isna().mean(),

        "acf_lag1": a[1],
        "acf_lag7": a[7],
        "pacf_lag1": pacf_lag1,
        "ljung_box_p": ljung_p,
        "adf_p": adf_p,
        "kpss_p": kpss_p,

        "trend_slope_per_day": pente.slope,
        "trend_pval": pente.pvalue,

        "seasonal_amplitude": seasonal.max() - seasonal.min(),

        "target_drift_pval": stats.ks_2samp(
            x[:len(x) // 2],
            x[len(x) // 2:]
        ).pvalue,

        "n_days_above_25": (x > 25).sum(),
        "n_days_above_50": (x > 50).sum(),

        "pct_days_above_25": 100 * (x > 25).mean(),
        "pct_days_above_50": 100 * (x > 50).mean(),

        "cv_strategy_recommended": "temporal",
        "gap_between_train_test_sec": 86400
    })

tempo = pd.DataFrame(tempo)

tempo.to_csv(
    OUT / "Exploratory_temporal.csv",
    index=False
)

if not tempo.empty:

    display(
        tempo[
            [
                "node_id",
                "n_records",
                "pct_missing_intervals",
                "acf_lag1",
                "acf_lag7",
                "ljung_box_p",
                "adf_p",
                "kpss_p"
            ]
        ].round(3).head()
    )

else:
    print("Aucune station avec suffisamment de données.")

derive = df.groupby(
    df[DATE].dt.year
)[TARGET].agg(
    mediane="median",
    iqr=lambda s: s.quantile(.75) - s.quantile(.25),
    pct_sup_25=lambda s: 100 * (s > 25).mean(),
    pct_sup_50=lambda s: 100 * (s > 50).mean()
)

display(derive.round(2))

ll = {
    c: [
        df.groupby(STATION).apply(
            lambda g: g[TARGET].corr(g[c].shift(k))
        ).median()
        for k in range(4)
    ]
    for c in CLES[1:]
}

print("Lead-lag correlations :")

display(
    pd.DataFrame(
        ll,
        index=["k=0", "k=1", "k=2", "k=3"]
    ).T.round(2)
)

# ============================================================
# ETAPE 19 : FEATURE ENGINEERING SANS FUITE
# ============================================================

df = df.sort_values([STATION, DATE])

grille = []

for st, g in df.groupby(STATION):

    g = g.set_index(DATE).sort_index()

    full_dates = pd.date_range(
        g.index.min(),
        g.index.max(),
        freq="D"
    )

    g = g.reindex(full_dates)

    g[STATION] = st

    grille.append(g)

data = pd.concat(grille)

data = (
    data.rename_axis(DATE)
    .reset_index()
)

grp = data.groupby(STATION)

# Lags
for k in [1, 2, 3]:
    data[f"pm25_lag{k}"] = grp[TARGET].shift(k)

for c in PRED:
    data[f"{c}_lag1"] = grp[c].shift(1)

# Rolling features
data["pm25_moy3"] = grp[TARGET].transform(
    lambda s: s.shift(1).rolling(3).mean()
)

data["pm25_moy7"] = grp[TARGET].transform(
    lambda s: s.shift(1).rolling(7).mean()
)

data["pm25_std7"] = grp[TARGET].transform(
    lambda s: s.shift(1).rolling(7).std()
)

data["pm25_diff1"] = (
    data["pm25_lag1"] - data["pm25_lag2"]
)

# Temporal features
jour = data[DATE].dt.dayofyear

data["jour_sin"] = np.sin(
    2 * np.pi * jour / 365.25
)

data["jour_cos"] = np.cos(
    2 * np.pi * jour / 365.25
)

data["semaine_sin"] = np.sin(
    2 * np.pi * data[DATE].dt.dayofweek / 7
)

data["semaine_cos"] = np.cos(
    2 * np.pi * data[DATE].dt.dayofweek / 7
)

FEATURES = (
    [f"pm25_lag{k}" for k in [1, 2, 3]]
    + [f"{c}_lag1" for c in PRED]
    + [
        "pm25_moy3",
        "pm25_moy7",
        "pm25_std7",
        "pm25_diff1",
        "jour_sin",
        "jour_cos",
        "semaine_sin",
        "semaine_cos"
    ]
)

data = data[
    data[TARGET].notna()
].reset_index(drop=True)

data["row_id"] = np.arange(len(data))

corr_cible = data[FEATURES].corrwith(
    data[TARGET]
).abs()

assert TARGET not in FEATURES

assert corr_cible.max() < 0.999, "FUITE DE DONNEES DETECTEE !"

print("Test anti-fuite OK")
print("Nombre de features :", len(FEATURES))
print(FEATURES)

# ============================================================
# ETAPE 20 : DEVELOPMENT / TEST
# ============================================================

TEST_START = pd.Timestamp("2025-01-01")

dev = data[
    data[DATE] < TEST_START
].reset_index(drop=True)

test = data[
    data[DATE] >= TEST_START
].reset_index(drop=True)

print("Développement :", len(dev), "lignes")
print("Test futur :", len(test), "lignes")

if dev.empty or test.empty:
    raise ValueError(
        "Séparation impossible : vérifie les dates du dataset "
        "et la valeur TEST_START."
    )

# ============================================================
# ETAPE 21 : CROSS VALIDATION
# ============================================================

def sauver(nom, folds):

    contenu = [
        {
            "fold": i,
            "train": dev.loc[tr, "row_id"].tolist(),
            "val": dev.loc[va, "row_id"].tolist()
        }
        for i, (tr, va) in enumerate(folds)
    ]

    with open(OUT / nom, "w") as f:
        json.dump(contenu, f, indent=2)


dates = np.sort(dev[DATE].unique())

if len(dates) < 7:
    raise ValueError("Pas assez de dates pour 5 folds temporels.")

f_temp = [
    (
        dev.index[dev[DATE].isin(dates[a])],
        dev.index[dev[DATE].isin(dates[b])]
    )
    for a, b in TimeSeriesSplit(
        n_splits=5,
        gap=1
    ).split(dates)
]

sauver("folds_temporal.json", f_temp)

if dev[STATION].nunique() >= 5:

    f_spat = list(
        GroupKFold(n_splits=5).split(
            dev,
            groups=dev[STATION]
        )
    )

    sauver("folds_spatial.json", f_spat)

else:
    print("Moins de 5 stations : validation spatiale ignorée.")

f_rand = list(
    KFold(
        n_splits=5,
        shuffle=True,
        random_state=42
    ).split(dev)
)

sauver("folds_standard.json", f_rand)

for i, (tr, va) in enumerate(f_temp):

    print(
        f"Fold temporel {i} : "
        f"train {dev.loc[tr, DATE].min().date()} "
        f"-> {dev.loc[tr, DATE].max().date()} | "
        f"val {dev.loc[va, DATE].min().date()} "
        f"-> {dev.loc[va, DATE].max().date()}"
    )

# ============================================================
# ETAPE 22 : PREPROCESSING PIPELINE
# ============================================================

pipeline = Pipeline([
    (
        "imputation",
        SimpleImputer(
            strategy="median",
            add_indicator=True
        )
    ),
    (
        "normalisation",
        StandardScaler()
    )
])

tr, va = f_temp[0]

X_train = pipeline.fit_transform(
    dev.loc[tr, FEATURES]
)

X_val = pipeline.transform(
    dev.loc[va, FEATURES]
)

print("Fold 0 :")
print("X_train :", X_train.shape)
print("X_val :", X_val.shape)
print("NaN restants :", np.isnan(X_val).sum())

# ============================================================
# ETAPE 23 : ARIMA DIAGNOSTICS
# ============================================================

p_adf = adfuller(serie_clean)[1]

p_kpss = kpss(
    serie_clean,
    nlags="auto"
)[1]

d = 0 if (
    p_adf < 0.05 and p_kpss > 0.05
) else 1

print(
    f"ADF p = {p_adf:.3f} | "
    f"KPSS p = {p_kpss:.3f} "
    f"-> ordre de différenciation d = {d}"
)

acf_lag7_median = (
    tempo["acf_lag7"].median()
    if not tempo.empty else np.nan
)

print(
    f"ACF lag 7 médiane = {acf_lag7_median:.2f} "
    "-> examiner une saisonnalité hebdomadaire (s=7)"
)

# ============================================================
# FINAL SAVING
# ============================================================

data.to_csv(
    OUT / "data_features.csv",
    index=False
)

feature_config = {
    "features": FEATURES,
    "target": TARGET,
    "test_start": "2025-01-01",
    "d_arima": int(d)
}

with open(OUT / "feature_config.json", "w") as f:
    json.dump(feature_config, f, indent=2)

print("\n==========================================")
print("PARTIE A TERMINEE !")
print("==========================================")

print("\nFichiers sauvegardés dans :", OUT)

for file in OUT.iterdir():
    print("-", file.name)
```
